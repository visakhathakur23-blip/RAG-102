### LANGCHAIN +TYPESENSE +ANthropics +RAG APPLICATION

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import Typesense
from langchain_text_splitters import CharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_anthropic import ChatAnthropic
```

```python
from dotenv import load_dotenv
import os

load_dotenv()

ENTROPY_API_KEY = os.getenv("ENTROPY_API_KEY")
```

```python
loader = TextLoader("test.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
docs = text_splitter.split_documents (documents)

embeddings = HuggingFaceEmbeddings()
```

```python
docsearch = Typesense.from_documents(
    docs,
    embeddings,
    typesense_client_params={
        'host':'as2prk8h9fbo3760p-1.a1.typesense.net',
        'port':'443',
        'protocol':'https',
        'typesense_api_key':'mI25noH6KABCycrLRGwSxVmIjjvLCHaD',
        'typesense_collection_name': 'lan-chain'
    },

)



```

```python
query = " Challenges and Ethical Dilemmas "
found_docs = docsearch.similarity_search(query)
print(found_docs[0].page_content)
```
